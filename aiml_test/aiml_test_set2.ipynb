#Neural Networks Q2

```python
import torch
import torch.nn as nn
```

```python
#nn.Module: -this is basically the base class or the building block of all the neural networks.

class SingleLayer(nn.Module):

  def __init__(self):
    super().__init__()

    #this basically states that it will take 2 inputs which will be then converted to 4 outputs.
    self.linear = nn.Linear(2,4)

    #ReLU is an activation function which will convert all the negative values to 0
    self.reLU = nn.ReLU()

  #this is a mandatory function which basically defines what our network will calculate during a forward pass, basically takes the input and returns the output
  def forward(self,x):
    x = self.linear(x)
    x = self.reLU(x)

    #returns the generated output
    return x


#first we instanciate the class
model = SingleLayer()

#our input tensor
ip_tensor = torch.tensor([1.0,-1.0])

output = model(ip_tensor)

print(output)
```

#LLMs Q1

Attention is a very interesting concept, so before that we should understand what the problem is and how attention solves it.
like if we have a sentence:-
##The cat sat on the mat
without attention if we try to figure out what does the word 'sat' represent then at that time it wont have any contextual meaning
##How does Attention solves the problem

Attention basically provides contextual meaning to each and every word, so what it basically does is that, every word has 4 vectors(embedding,query,key,value) and attention for every words has a query which then looks for the relevant key for every other word and then infuses the embedding of that particular word with the values of the relevant keys, so it provides the contextual meaning to that word

#LLMs Q2

```python
from transformers import GPT2Tokenizer, GPT2LMHeadModel

tokenizer = GPT2Tokenizer.from_pretrained("openai-community/gpt2") #this is basically used to convert the raw text into token ids
model = GPT2LMHeadModel.from_pretrained("openai-community/gpt2") #this is basically the LLM head which the last layer of the network

inputs = tokenizer("Once upon a time", return_tensors="pt")
```

```python
#when do_sample is set to true then at that time the random sampling is enabled and it can provide more creativity.

generation_output = model.generate(**inputs, do_sample=True)

#convert the generated output back to raw text
decoded_output = tokenizer.decode(generation_output[0])
print(decoded_output)
```

```python
#when do_sample is set to false then at that time a greedy decoding strategy is used and model will select the tokens with highest probability leading to predictable answers

generation_output = model.generate(**inputs, num_beams=4, do_sample=False)
decoded_output = tokenizer.decode(generation_output[0])
print(decoded_output)
```
